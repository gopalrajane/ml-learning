```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import make_regression
```

```python
x,y = make_regression(n_samples=4, n_features=1, n_informative=1, n_targets=1,noise=80,random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
```

```python
# Applying Ordinal Least Square

from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()

reg.fit(x,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red')
```

```python
# applying gradient descent assuming m is constant m = 78.35
# assume intercent b = 0
```

```python
y_pred = ((78.5 * x) + 100).reshape(4)
```

```python
# Plot Graph

plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label='OLS')
plt.plot(x,y_pred,color='green',label='b = 0')
plt.legend()
```

```python
m = 78.35
b = 100

loss_slope = -2 * np.sum(y - m*x.ravel() - b)
loss_slope
```

```python
# take learning rate = 0.1

lr = 0.1

step_size = loss_slope*lr
step_size
```

```python
# calculating new intercepts
b = b - step_size
b
```

```python
y_pred1 = ((78.35 * x) + b).reshape(4)

plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label='OLS')
plt.plot(x,y_pred,color='lightblue',label='b = 0'.format(b))
plt.plot(x,y_pred1,color='green',label='2nd b')
plt.legend()
```

```python
# 2nd Iteration 

loss_slope = -2 * np.sum(y-m*x.ravel() - b)
loss_slope
```

```python
step_size = loss_slope * lr
```

```python
step_size
```

```python
b = b - step_size
```

```python
y_pred2 = ((78.35 * x) + b).reshape(4)
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label='OLS')
plt.plot(x,y_pred,color='lightblue',label='b = 0'.format(b))
plt.plot(x,y_pred1,color='green',label='2nd b')
plt.plot(x,y_pred2,color='blue',label='3rd b')
plt.legend()
```

```python
# 3rd Iteration 

loss_slope = -2 * np.sum(y - m*x.ravel() - b)
loss_slope
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred3 = ((78.35 * x) + b)
```

```python
plt.plot(x,reg.predict(x),color='red',label='OLS')
plt.plot(x,y_pred,color='lightblue',label='b = 0'.format(b))
plt.plot(x,y_pred1,color='green',label='2nd b')
plt.plot(x,y_pred2,color='blue',label='3rd b')
plt.plot(x,y_pred3,color='yellow')
plt.legend()
```

```python
# create a loop for all operation 

b = 100
m = 78.35
lr = 0.01


epochs = 100

for i in range(epochs):
    loss_slope = -2 * np.sum(y - m*x.ravel() - b)
    step_size = loss_slope * lr
    b = b - step_size

    y_pred = m * x + b

    plt.plot(x,y_pred)

plt.scatter(x,y)
```
