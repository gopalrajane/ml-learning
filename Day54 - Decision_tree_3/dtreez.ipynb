# 3. Horizontal Decision Tree

```python
import os

os.environ["PATH"] = r"C:\Program Files\Graphviz\bin;" + os.environ["PATH"]

iris = load_iris()
X = iris.data
y = iris.target

clas = DecisionTreeClassifier(max_depth=2)
clas.fit(X, y)


viz = model(
    model=clas,
    X_train=X,
    y_train=y,
    feature_names=iris.feature_names,
    class_names=iris.target_names
)

viz.view()
```

# 5. Show node number

```python
clas = DecisionTreeClassifier(max_depth=2)
clas.fit(X, y)

viz = model(
    model=clas,
    X_train=X,
    y_train=y,
    feature_names=iris.feature_names,
    class_names=iris.target_names
)


viz.view(
    orientation="LR",          
    show_node_labels=True,  
    histtype="barstacked",     
    scale=1.5                  
)
```
